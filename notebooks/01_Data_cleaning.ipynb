```python
# so in the UCI Machine learning they explicitly mention the education are only in between 0,1,2 and 4
# So to avoid any miss-happens we are making 0,5,6,7,8.--> 4
df['EDUCATION'] = df['EDUCATION'].replace({0: 4, 5:4, 6:4})

# verifying 
print(df['EDUCATION'].value_counts().sort_index())
```

```python
# Check what happened
print(df['EDUCATION'].value_counts())
print(df['EDUCATION'].isnull().sum())
```

```python
print(df.columns.tolist())
```

```python
# for PAY columns only valid values are 
for col in ['PAY_0', 'PAY_2','PAY_3','PAY_4','PAY_5','PAY_6']:
    print(col, df[col].unique())
# in UCI they explicitly mention -1=pay_duly, =1, 1 month delay...2 for 2 month...so on til 9 
```

```python
pay_cols = ['PAY_0','PAY_2','PAY_3','PAY_4','PAY_5','PAY_6']
df[pay_cols] = df[pay_cols].replace({-2: -1, 0:-1})
# here we replce -2 with -1 and 0 with -1
```

```python
for col in ['BILL_AMT1', 'BILL_AMT2','BILL_AMT3','BILL_AMT4', 'BILL_AMT5','BILL_AMT6']:
    print(col, 'Min:', df[col].min(), 'Max:', df[col].max())
# so here the are allow cause the credit balance are overpaid
```

```python
for col in ['PAY_AMT1','PAY_AMT2','PAY_AMT3',
            'PAY_AMT4','PAY_AMT5','PAY_AMT6']:
    print(col, 'Min:', df[col].min(), 'Max:', df[col].max())
    # so there's no negative value so it's already clean
```

```python
for col in ['PAY_AMT1','PAY_AMT2','PAY_AMT3',
            'PAY_AMT4','PAY_AMT5','PAY_AMT6']:
    print(col, 'Min:', df[col].min(), 'Max:', df[col].max())
```

```python
cat_cols = ['EDUCATION', 'PAY_0', 'PAY_2', 'PAY_3', 
            'PAY_4', 'PAY_5', 'PAY_6', 'DEFAULT']

df[cat_cols] = df[cat_cols].astype('category')
```
